````python
EXPECTED_KEYS = {
    "sentiment": str,
    "summary": str,
    "positive_points": list,
    "negative_points": list,
    "complaint_category": list,
    "needs_human_followup": bool
}
````

````python
ALLOWED_SENTIMENTS = ["positive", "neutral", "negative", "mixed"]

ALLOWED_CATEGORIES = [
    "food_quality",
    "service",
    "price",
    "waiting_time",
    "delivery",
    "cleanliness",
    "noise",
    "wrong_item",
    "other"
]
````

````python
def build_prompt(reviews:str)->list:
    return [{'role':'user', 'content':f"""
    你是一個評論分析器。請分析以下評論，並只輸出 JSON，不要使用 Markdown，不要加任何解釋。

    評論：
    {reviews}

    請輸出以下格式：

    {{
    "sentiment": "positive | neutral | negative | mixed",
    "summary": "一句話摘要",
    "positive_points": ["正面重點1", "正面重點2"],
    "negative_points": ["負面重點1", "負面重點2"],
    "complaint_category": ["food_quality | service | price | waiting_time | delivery | cleanliness | noise | wrong_item | other"],
    "needs_human_followup": true
    }}
    """}]

def call_model(prompt:list, max_new_tokens:int=512)->str:
    output = pipe(
        prompt,
        max_new_tokens=max_new_tokens,
        do_sample=False
    )
    return output[0]['generated_text'][-1]['content']
````

````python
import json

def parse_json_direct(raw:str):
    return json.loads(raw)
````

````python
prompt = build_prompt(reviews[0])
raw = call_model(prompt)

print('Raw Output:')
print(raw)

print('-'*20)
print('Parsed Output')

parsed = parse_json_direct(raw)
print(parsed)
````

上面的解析有可能會出現 JSONDecodeError
```
```json   
{ ... }
```

或：
```
```text  
以下是 JSON：  
{ ... }
```

````python
import re

def extract_json_block(text:str)->str:
    text = text.strip()

    # 移除 Markdown code fence
    text = re.sub(r'^```json\s*', '', text)
    text = re.sub(r'^```\s*', '', text)
    text = re.sub(r'\s*```&', '', text)

    # 如果前後有解釋文字，抓第一個 { 到最後一個 }
    start = text.find('{')
    end = text.rfind('}')

    if start == -1 or end == -1 or end <= start:
        raise ValueError('No JSON object found')

    return text[start:end+1]

def parse_json_robust(raw:str):
    json_text = extract_json_block(raw)
    return json.loads(json_text)
````

````python
def validate_basic(data:dict):
    errors = []

    for key, expected_type in EXPECTED_KEYS.items():
        if key not in data:
            errors.append(f'Missing key: {key}')
        elif not isinstance(data[key],expected_type):
            errors.append(
                f'Wrong type for {key}: expected {expected_type.__name__}, got {type(data[key]).__name__}'
            )

        if 'sentiment' in data and data['sentiment'] not in ALLOWED_SENTIMENTS:
            errors.append(f'Invalid sentiment: {data['sentiment']}')
        if 'complaint_category' in data and isinstance(data['complaint_category'],list):
            for c in data['complaint_category']:
                if c not in ALLOWED_CATEGORIES:
                    errors.append(f'Invalid complaint_category: {data['complaint_category']}')
    return errors

data = parse_json_robust(raw)
errors = validate_basic(data)
if errors:
    print('Validation Error:')
    for error in errors:
        print('-', error)
else:
    print('Valid output')
````

````python
results = []

for i, review in enumerate(reviews):
    prompt = build_prompt(review)
    raw = call_model(prompt)

    record = {
        "index": i,
        "review": review,
        "raw": raw,
        "parse_success": False,
        "validation_success": False,
        "errors": []
    }

    try:
        data = parse_json_robust(raw)
        record["parse_success"] = True

        errors = validate_basic(data)
        record["errors"] = errors
        record["validation_success"] = len(errors) == 0
        record["parsed"] = data

    except Exception as e:
        record["errors"] = [str(e)]

    results.append(record)
````

````python
parse_success_rate = sum(r["parse_success"] for r in results) / len(results)
validation_success_rate = sum(r["validation_success"] for r in results) / len(results)

print("Parse success rate:", parse_success_rate)
print("Validation success rate:", validation_success_rate)
````
